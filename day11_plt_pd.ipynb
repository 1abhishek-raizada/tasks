```python
gas=pd.read_csv('gas_prices.csv')
plt.figure(figsize=(6,4),dpi=100)
plt.title('Gas prices in US $')
#plt.plot(gas.Year,gas.Canada,label='Canada')
#plt.plot(gas.Year,gas.USA,label='United States')
#plt.plot(gas.Year,gas.UK,label='United Kingdom')
for country in gas:
    if country != 'Year':
        plt.plot(gas.Year,gas[country],label=f'{country}')

plt.xticks(gas.Year[::3])
plt.xlabel('Years')
plt.ylabel('price in Dollars')
plt.legend(loc='upper left',bbox_to_anchor=(1,1))
plt.show()
```

LOADING ANOTHER DATASET

```python
fifa = pd.read_csv('fifa_data.csv')
```

```python
fifa.head(10)
```

```python
#histogram
plt.figure(figsize=(4,4),dpi=100)
bins=np.arange(40,110,10)
plt.xticks(bins)
plt.hist(fifa.Overall,bins=bins,color='#cbaafd')
plt.xlabel('skill level')
plt.ylabel('number of players')
plt.title('FIFA player level Distribution')
plt.show()
```

```python
#PIE CHART

left=fifa.loc[fifa['Preferred Foot']=='Left'].count()[0]
right=fifa.loc[fifa['Preferred Foot']=='Right'].count()[0]
labels=['Left Foot','Right Foot']
plt.pie([left,right],labels=labels,autopct='%.2f %%')
plt.show()
```

```python

fifa.Weight=[int(x.strip('lbs')) if type(x)==str else x for x in fifa.Weight]

light=fifa.loc[fifa.Weight < 125].count()[0]
light_medium=fifa.loc[(fifa.Weight >= 125) & (fifa.Weight <150)].count()[0]
medium=fifa.loc[(fifa.Weight >= 150) & (fifa.Weight < 175)].count()[0]
medium_heavy=fifa.loc[(fifa.Weight >= 175) & (fifa.Weight < 200)].count()[0]
heavy=fifa.loc[fifa.Weight >= 200].count()[0]



```

```python

weights=[light,light_medium,medium,medium_heavy,heavy]

labels1=['Under 125','125-150','150-175','175-200','Over 200']
plt.style.use('ggplot')
explode=[1,0,0,0,.4]
plt.pie(weights,labels=labels1,autopct='%.2f %%',pctdistance=0.8,explode=explode)
plt.title('FIFA player skill Distribution')
plt.show()
```
