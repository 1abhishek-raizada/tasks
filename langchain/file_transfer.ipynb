```python
import shutil
import os


```

```python
name=input("Enter the output folder name:")
```

```python
source=os.getcwd()
another_name=input("enter input folder:")
inputs=another_name
input_folder=os.path.join(source,inputs)


destination=os.path.join(source,name)
print(destination)

os.makedirs(destination,exist_ok=True)
print(os.getcwd())
```

```python
from pathlib import Path
```

```python
for files in os.listdir(input_folder):
    print(files)
    inputt=os.path.join(input_folder,files)

    c=Path(inputt).suffix
    
    folders=os.path.join(destination,c)
    os.makedirs(folders,exist_ok=True)

    destinationf=os.path.join(folders,files)
    shutil.copy(inputt,destinationf)
```
