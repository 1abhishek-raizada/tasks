```python
from langchain.prompts import PromptTemplate
from langchain_aws import Bedrock
import boto3

# Setup Bedrock client
session = boto3.Session(region_name="us-east-1")
llm = Bedrock(
    client=session.client("bedrock-runtime"),
    model_id="mistral.mistral-7b-instruct-v0:2",
    model_kwargs={"temperature": 0.1, "max_tokens": 512}
)

# (Optional) test Bedrock is working
# print(llm.invoke("Where is Tokyo?"))

# Define the prompt
template = """
Previous conversation:
{messages}

User says:
{input}

Reply as a helpful assistant.
"""

prompt = PromptTemplate(
    input_variables=["messages", "input"],
    template=template
)

# Build LCEL chain
chain = prompt | llm

# Example inputs
query = {
    "messages": "User: Hello, where is Tokyo?\nAssistant: Tokyo is in Japan.",
    "input": "What can I eat there?"
}

# Invoke the chain
gen_ans = chain.invoke(query)

# gen_ans will likely be a string
response_text = gen_ans.strip() if isinstance(gen_ans, str) else getattr(gen_ans, "content", "").strip()

if not response_text:
    response_text = "Sorry, I don't have knowledge of it. Please try something else."

return {
    "status_code": 200,
    "message": response_text,
    "total_cost": 0,
    "total_token": 0
}
```
