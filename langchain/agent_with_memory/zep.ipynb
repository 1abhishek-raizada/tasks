```python
user_id="abhishekraizada"
new_user=client.user.add(
    user_id=user_id,
    email="abhiraizada@gmail.com",
    first_name="abhishek",
    last_name="raizada",
)
```

```python
import uuid

#generating unique session id
session_id=uuid.uuid4().hex

client.memory.add_session(
    session_id=session_id,
    user_id=user_id,
)
```

```python
from zep_cloud.types import Message

messages=[
    Message(
        role='Abhishek',
        content='Hi, my name is Abhishek Raizada',
        role_type="user",
    ),
    Message(
        role="AI assistant",
        content="Hello Abhishek, Nice to meet you",
        role_type='assistant',
    )
]
```

```python
client.memory.add(session_id,messages=messages)
```

```python
memory=client.memory.get(session_id=session_id)

context_string=memory.context
print(context_string)

recent_messages=memory.messages

for msg in recent_messages:
    print(f"{msg.role}:{msg.content}")
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-1.5-flash-latest",temperature=0)
```

```python
bot_name="Jarvis"
user_name="Abhishek"
user_id=user_name + str(uuid.uuid4())[:4]
session_id=str(uuid.uuid4())

client.user.add(
    user_id=user_id,
    email=f"{user_name}@raizada.com",
    first_name=user_name,
    last_name="Raizada",
)

client.memory.add_session(
    user_id=user_id,
    session_id=session_id,
)
```

```python
from langchain_core.messages import SystemMessage,HumanMessage
```

```python
from zep_cloud import FileNotFoundError
```

```python
user_id = "user123"
user_name = "abhishek"
new_session_id = str(uuid.uuid4())
def chatting(user_input: str):
    try:
        client.memory.get_session(session_id=session_id)
    except FileNotFoundError:
        client.memory.add_session(user_id=user_id, session_id=session_id)
    client.memory.add(
        session_id=session_id,
        messages=[Message(role_type="user", role=user_name, content=user_input)],
    )

    memory = client.memory.get(session_id=session_id)
    past_context = memory.context or "No context present"

    messages = [
        SystemMessage(content="You are a helpful, friendly customer support agent."),
        HumanMessage(content=f"Memory: {past_context}"),
        HumanMessage(content=user_input),
    ]

    response = llm.invoke(messages)
    print(response.content)
```
