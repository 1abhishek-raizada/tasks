```python
def function1(input1):
    return input1 + "from first function"

def function2(input2):
    return input2 + "and this is second function"
```

```python
from langgraph.graph import Graph
```

```python
workflow1=Graph()
```

```python
workflow1.add_node("func1",function1)
```

```python
workflow1.add_node("func2",function2)
```

```python
workflow1.add_edge("func1","func2")
```

```python
workflow1.set_entry_point("func1")
```

```python
workflow1.set_finish_point("func2")
```

```python
app1=workflow1.compile()
```

```python
from IPython.display import display
from PIL import Image
import io

try:
    img_data = app1.get_graph().draw_mermaid_png()  # should be raw PNG bytes
    img = Image.open(io.BytesIO(img_data))
    display(img)
except Exception as e:
    print("Could not render Mermaid diagram:", e)
```

```python
app1.invoke("hello this is abhishek " )
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings=GoogleGenerativeAIEmbeddings(model="models/embeddings-001")

from langchain_google_genai import ChatGoogleGenerativeAI

llm=ChatGoogleGenerativeAI(model="gemini-1.5-flash-latest")
```

```python
llm.invoke("hi").content
```

```python
def fun1(input):
    llm=ChatGoogleGenerativeAI(model="gemini-1.5-flash-latest")
    response=llm.invoke(input).content
    return response
```

```python
def fun2(input):
    upper_case=input.upper()
    return upper_case
```

```python
workflow2=Graph()
```

```python
workflow2.add_node("llm",fun1)
```

```python
workflow2.add_node("uppercase",fun2)
```

```python
workflow2.add_edge("llm","uppercase")
```

```python
workflow2.set_entry_point("llm")
```

```python
workflow2.set_finish_point("uppercase")
```

```python
app2=workflow2.compile()
```

```python
from IPython.display import display
from PIL import Image
import io

try:
    img_data = app2.get_graph().draw_mermaid_png()  # should be raw PNG bytes
    img = Image.open(io.BytesIO(img_data))
    display(img)
except Exception as e:
    print("Could not render Mermaid diagram:", e)
```

```python
app2.invoke("hello gemini")
```
